```python
# to split train and test data

from sklearn.model_selection import train_test_split

# Split training and testing data
x_train, x_test, y_train, y_test = train_test_split( x, y, test_size=0.2, random_state=0)

melb_model_obj = DecisionTreeRegressor()

#fit model
melb_model_obj.fit(x_train,y_train)

#to find output prediction
output_val_prediction = melb_model_obj.predict(x_test) # here predicted of input test data
print(mean_absolute_error(y_test,output_val_prediction ))
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error

# define function 
def get_mae(max_leaf_nodes,x_train, x_test, y_train, y_test):
    model_name = DecisionTreeRegressor(max_leaf_nodes = max_leaf_nodes, random_state=0)
    model_name.fit(x_train,y_train)
    predtion_test_data = model_name.predict(x_test)
    mean_abs_error = mean_absolute_error(y_test,predtion_test_data)
    return(mean_abs_error)
```

```python

for max_leaf_nodes in [5,50,500,5000]:
    new_mae = get_mae(max_leaf_nodes,x_train, x_test, y_train, y_test)
    print("Max of leaf node: %d \t\t mean absolute error:%d" %( max_leaf_nodes,new_mae))

    #here check good fit,it is placed bw underfit and overfit 
```

```python
from sklearn.ensemble import RandomForestRegressor
forest_model = RandomForestRegressor()
forest_model.fit(x_train,y_train)

melb_predict = forest_model.predict(x_test)
mean_absolute_error(y_test,melb_predict)
```
