```python
import os
import csv
# import pandas as pd
# import matplotlib.pyplot as plt

# Folder and file path
#joining the path
path ="C:\\Users\\Admin\\Desktop\\my python project"
myfile="mylog.log"
print(os.path.join(path,myfile))

# Read the log file safely
ip_list=[]
http_status_list=[]

with open(os.path.join(path,myfile),'r')as file:
    for each_line in file:
        item =each_line.split()
        if len(item) > 8:  # confirm 
            ip_list.append(item[3])          # IP
            http_status_list.append(item[8]) # HTTP status code

# Function to count frequency

def frequency(mylist):   
    mydict={} 
    for each in mylist:
        if each not in mydict:
            mydict[each]=1
        else:
            mydict[each]+=1
    return mydict

# Function to write dictionary to CSV
def write_to_csv(mydict, outfile, header1="Item", header2="Count"):
    with open(outfile, "w", newline="") as f:
        writer = csv.writer(f)

        writer.writerow(["S.No", header1, header2])

        for i, (key, value) in enumerate(mydict.items(), start=1):
            writer.writerow([i, key, value])

# Get frequency of IPs and status codes
ip_counts = frequency(ip_list)
status_counts = frequency(http_status_list)

# Write results to CSV
write_to_csv(ip_counts, os.path.join(path, "IP.csv"), "IP", "Frequency")
write_to_csv(status_counts, os.path.join(path, "httP.csv"), "HTTP_Status", "Count")

print("\nCSV files created successfully!")
```

```python

import os
import pandas as pd
import matplotlib.pyplot as plt


# File path
main_path = r"C:\Users\Admin\Desktop\heart details"
file_name = "heart.csv"

# Read heart disease file
data = pd.read_csv(os.path.join(main_path, file_name ))

# Extract columns
chol = data["chol"]
thalach = data["thalach"]
age = data["age"]

# Count values
chol_count = chol.value_counts()
thalach_count = thalach.value_counts()
age_count = age.value_counts()

# Save results to CSV
chol_count.to_csv("cholesterol_count.csv")
thalach_count.to_csv("thalach_count.csv")
age_count.to_csv("age_count.csv")

# Visualize data
plt.hist(chol, bins=10)
plt.title("Cholesterol Distribution")
plt.xlabel("Cholesterol")
plt.ylabel("Frequency")
plt.show()

plt.hist(thalach, bins=10)
plt.title("Maximum Heart Rate Distribution")
plt.xlabel("Maximum Heart Rate")
plt.ylabel("Frequency")
plt.show()

plt.hist(age, bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
import csv
import pandas as pd
import matplotlib.pyplot as plt
import os

#-------------- Create output folders -----------------------#
main_path= r"C:\Users\Admin\Desktop\heart details"
file = "heart.csv"
output_folder = os.path.join(main_path, "output")
chart_folder = os.path.join(output_folder, "charts")

os.makedirs(chart_folder, exist_ok=True)

# exist_ok=True — to prevent the error ,if already folder is occur

#------------------------ Read the CSV file--------------------------------------#

main_path= r"C:\Users\Admin\Desktop\heart details"
file = "heart.csv"

df = pd.read_csv(os.path.join(main_path, file))

print("Heart Disease Dataset Loaded Successfully!")
print()
print("Total rows:", len(df))
print()
print(df.head(5))


#------------------------  Extract the required columns----------------------------#
# to extract  Required columns 
chol = df["chol"]
thalach = df["thalach"]
age = df["age"]

# df["age"] ----only take the column value of age in data

#-------------------- to Calculate frequencies -------------------------------------#

chol_freq = chol.value_counts().sort_index()     #here sort is give ascending order
thalach_freq = thalach.value_counts().sort_index()
age_freq = age.value_counts().sort_index()

# -----------------------display  Frequency results------------------------------------#

print("\n---------------AGE FREQUENCY ----------------")
print(age_freq.head())
print("\n---------- CHOLESTEROL FREQUENCY -------------")
print(chol_freq.head())
print("\n------------- THALACH FREQUENCY --------")
print(thalach_freq.head())

#---------------------------  write frequency result in csv file-----------------------#
age_csv = os.path.join(output_folder, "age_frequency.csv")
chol_csv = os.path.join(output_folder, "chol_frequency.csv")
thalach_csv = os.path.join(output_folder, "thalach_frequency.csv")

age_freq.reset_index().to_csv(age_csv, index=False, header=["Age", "Frequency"])
chol_freq.reset_index().to_csv(chol_csv, index=False, header=["Cholesterol", "Frequency"])
thalach_freq.reset_index().to_csv(thalach_csv, index=False, header=["Thalach", "Frequency"])

#resert_index()-to change index value to normal column value
#index=False----to prevent not add extra cloumn

print("\nCSV files saved:")
print(age_csv)
print(chol_csv)
print(thalach_csv)

#-----------------to create Visualization function------------------------#

def plot_frequency(freq, title, xlabel, filename, top_n=None):
    # Show the most frequent values if top_n is specified
    if top_n is not None:
        freq = freq.nlargest(top_n).sort_index()  #sort is give ascending order

    plt.figure(figsize=(12, 5))
    plt.bar(freq.index.astype(str), freq.values)

    # here freq.index--disply x axis
    #.astype(str)---to change the string value in x axis
    #freq.values -- to display y axis freq 

    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Frequency")
    # plt.xticks(rotation=45, ha="right")
    plt.tight_layout()  # avoid overlap chart

    chart_path = os.path.join(chart_folder, filename)
    plt.savefig(chart_path)
    plt.show()
    plt.close()

    print("Chart saved:", chart_path)

# Age: show all unique ages
plot_frequency(age_freq,"Age Frequency","Age", "age_frequency.png")

# Cholesterol: show the 10 most frequent values
plot_frequency( chol_freq, "Cholesterol Frequency","Cholesterol", "chol_frequency.png",top_n=10)

# Thalach: show the 10 most frequent values
plot_frequency(thalach_freq,"Maximum Heart Rate Frequency","Thalach","thalach_frequency.png",top_n=10)

print("\nHeart dataset analysis completed successfully!")



```
