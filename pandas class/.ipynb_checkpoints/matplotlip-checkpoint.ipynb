## **sactter**  
  Used to study relationships between two variables.

  ***Function***: plt.scatter()


```python
x = np.array([1,2,3]) 
y = np.array([1,2,3]) 
plt.scatter(x,y,c = 'r',marker = "*",s=100) 
plt.show();
```

```python
x = np.array([1,2,3,4]) 
y = np.array([1,2,3,4]) 
color = ["blue","yellow",'red','g']
plt.scatter(x,y,c = color,marker = "*",s=100) 
plt.show();
```

### **Line grap**  
   Used to show trends over time or continuous data.  

   ***Function***: plt.plot()  

```python
x = np.array([1,3,4,5]) 
y = np.array([3,2,1,4]) 
color = ["blue","yellow",'red','g']

plt.plot(x,y,c='g',linestyle="dashed")
plt.scatter(x,y,c = color,marker = "*",s=100) 
plt.show();
```

### **Bar chart** 
Used to compare values across categories.

***Function***: plt.bar(), plt.barh()


```python
year = ["1990","2000","2010","2020","2030"] 
lang = ["None","Java","C","C++","Python"]
city=["chennai","cbe",'erode','kerala','none']
c = ['Green','darkorange','deeppink','violet',"red"]

plt.bar(year,lang,city,color=c,align='edge',widht=1)
plt.show();
```

```python
import matplotlib.pyplot as plt

year = ["1990", "2000", "2010", "2020", "2030"]
lang = ["None", "Java", "C", "C++", "Python"]
city = ["chennai", "cbe", "erode", "kerala", "none"]
c = ['Green', 'darkorange', 'deeppink', 'violet', 'red']

plt.bar(year, [1, 2, 3, 4, 5], color=c, align='edge', width=1)
plt.show()
```

### **Histograms:**
   Used to show the frequency distribution of continuous numerical data.  
  ***functions***:plt.hist()

```python
# frquency distribution 
#bule color chart is mention bins


np.random.seed(2) 
a = np.random.normal(50,1.5,100)  # 100=range value number,50=reak number of 100,2=distribute
plt.hist(a,color='pink',bins=15,edgecolor="black")
plt.show();
```

### **Pie Chart**  
   Used to display proportions or percentages.

  ***Function***: plt.pie()

```python
# autopct — automatically displays the percentage of each slice.

# %1.1f — formats the number as a floating-point value with 1 digit after the decimal point.

# %% — displays the actual % symbol.

lang = ["Python","Java","C++","C"]   
votes= [90,75,50,40]  #
exp = [0.1,0,0.3,0]
c = ['Green','darkorange','deeppink','violet']



plt.pie(votes,labels = lang,explode=exp,colors=c,autopct='%1.1f%%',shadow=True)  #total color change=colors
plt.show();
```

### **Add titile**  
plt.tittle()

```python
years=["2000","2010","2020","2030"]
income = [200,400,600,800]

plt.title("Daily income of employee")
plt.xlabel("Years")
plt.ylabel("income")
plt.plot(years,income,c="r")
plt.show();
```

### **legend**

```python
# Stack market
stockA = [1,2,3,4,5]
stockB = [3,4,5,3,2]
stockC = [6,4,2,1,3,4,5]

plt.title("Stock market")

plt.plot(stockA,label="stock1",marker="p")
plt.plot(stockB,label="stock2",marker="o")
plt.plot(stockC,label="stock1",marker="s")

plt.legend()
plt.show();

```

## **subplots comparing (age vs chol),(age vs thalach )**

```python


fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(12,8))

#subplot=1
scatter = ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
#customise the plot
ax[0].set(title ='Age vs Cholestrol',xlabel="Age",ylabel="Cholestrol") 
ax[0].legend(*scatter.legend_elements(),title="Target")   #*->args

#add mean line
ax[0].axhline(df["chol"].mean(),color="r",linestyle="dashed") #here chol meand value


#age vs thalach
scatter=ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="viridis")
#customise the plot
ax[1].set(title='Age vs Thalach',xlabel="Age",ylabel="Thalach")
ax[1].legend(*scatter.legend_elements(),title="Target")
ax[1].axhline(df["thalach"].mean(),color="r",linestyle="dashed")

plt.subplots_adjust(hspace=0.4)  #adjust the subplot
plt.show();
```

```python
# image  

#Create Plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot Data
scatter = ax.scatter(df['age'], df['chol'], c=df['target'],cmap="winter")

# Customize the plot
ax.set(title="Heart Disease and Cholesterol Levels", xlabel="Age",ylabel="Cholesterol")
ax.legend(*scatter.legend_elements(), title='Target')
# Add a meanline
ax.axhline(df['chol'].mean(), linestyle='dashed');

#image save
plt.savefig("Probability_of_heart_attack.png")
```

## **subplots comparing (age vs chol),(age vs thalach ),(age vs fbs),(ages vs trestbps)**

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(12,8))

#subplot=1
scatter = ax[0,0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter")
#customise the plot
ax[0,0].set(title ='Age vs Cholestrol',xlabel="Age",ylabel="Cholestrol") 
ax[0,0].legend(*scatter.legend_elements(),title="Target")   #*->args
#add mean line
ax[0,0].axhline(df["chol"].mean(),color="r",linestyle="dashed") #here chol meand value


#age vs thalach
scatter=ax[0,1].scatter(df["age"],df["thalach"],c=df["target"],cmap="viridis")
#customise the plot
ax[0,1].set(title='Age vs Thalach',xlabel="Age",ylabel="Thalach")
ax[0,1].legend(*scatter.legend_elements(),title="Target")
ax[0,1].axhline(df["thalach"].mean(),color="r",linestyle="dashed")


#age vs fbs(fasting blood sugar)
scatter=ax[1,0].scatter(df["age"],df["fbs"],c=df["target"],cmap="plasma")
ax[1,0].set(title="Age vs fasting blood sugar",xlabel="Age",ylabel="fbs")
ax[1, 0].legend(*scatter.legend_elements(), title="Target")
ax[1, 0].axhline(df["fbs"].mean(),color="b", linestyle="dashed")

# 4. Age vs Trestbps
scatter = ax[1, 1].scatter(df["age"], df["trestbps"],c=df["target"], cmap="plasma")
ax[1, 1].set(title="Age vs Trestbps",xlabel="Age", ylabel="Trestbps")
ax[1, 1].legend(*scatter.legend_elements(), title="Target")
ax[1, 1].axhline(df["trestbps"].mean(),color="b", linestyle="dashed")

# Adjust spacing and display
plt.subplots_adjust(hspace=0.3)
plt.show()

```
