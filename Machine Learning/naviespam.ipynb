In the SMS Spam Collection Dataset,  
our goal is to predict whether an SMS message is Spam or Ham (non-spam message) using the Naive Bayes algorithm.

v1 is output
v2 is input

```python


import pandas as pd

#encoding="latin-1" – Text encoding problem avoid panna use pannuvom.(Reads many Western European characters)
spam_data =pd.read_csv("C:\\techpanda\\course\\ML\\data\\navie bayas\\spam.csv",  encoding="latin-1")

spam_data.head(10)
```

```python
spam_data .info()
```

```python
spam_data .isna().sum()
```

```python
# set inuput and output
input_data = spam_data["v2"]
print(input_data.head(10))
```

```python
output_data = spam_data["v1"]
print(output_data.head(10))
```

```python
#split test and train

from sklearn.model_selection import train_test_split

input_data_train,input_data_test,output_data_train,output_data_test = train_test_split(input_data,output_data,test_size=0.2)
```

```python
#TfidfVectorizer is used to convert text data into numerical values
from sklearn.feature_extraction.text import TfidfVectorizer

#Learns from numerical data
from sklearn.naive_bayes import MultinomialNB

# make_pipeline() is used to combine multiple preprocessing and machine learning steps into a single object.
from sklearn.pipeline import make_pipeline
```

```python
#  Create model
object_model = make_pipeline(TfidfVectorizer(),MultinomialNB())
```

```python
# Train model
object_model.fit(input_data_train, output_data_train)
```

```python
# predict  input test data
predict_sms = object_model.predict(input_data_test)
predict_sms
```

```python
print(input_data_train.shape)
print(input_data_test.shape)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(output_data_test,predict_sms)
```

```python
pd.DataFrame(confusion_matrix(output_data_test,predict_sms),
             columns=["predicted ham","predicted spam"],
             index=["Actual ham","Actual spam"])
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(output_data_test,predict_sms) # actual output,predicted output
accuracy
```
