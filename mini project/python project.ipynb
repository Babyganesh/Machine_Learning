```python
# Read the log file safely
ip_list=[]
http_status_list=[]

with open(os.path.join(path,myfile),'r')as file:
    for each_line in file:
               parts =each_line.split()
               if len(parts) > 8:  # ensure enough fields exist
                           ip_list.append(parts[3])          # IP (check position in your log)
                           http_status_list.append(parts[8]) # HTTP status code

print(ip_list)
print(http_status_list)
```

```python
# Function to count frequency

def frequency(mylist):   
    mydict={} 
    for each in mylist:
        if each not in mydict:
            mydict[each]=1
        else:
            mydict[each]+=1
    return mydict

# Function to count frequency
# def frequency(mylist):
#     mydict = {}
#     for item in mylist:
#         mydict[item] = mydict.get(item, 0) + 1
#     return mydict
```

```python
# Function to write dictionary to CSV
def write_to_csv(mydict, outfile, header1="Item", header2="Count"):
    with open(outfile, "w", newline="") as f:
        writer = csv.writer(f)

        writer.writerow(["S.No", header1, header2])

        for i, (key, value) in enumerate(mydict.items(), start=1):
            writer.writerow([i, key, value])
```

```python
# Get frequency of IPs and status codes
ip_counts = frequency(ip_list)
status_counts = frequency(http_status_list)

# Print results
print("Unique IPs and their counts:")
print(ip_counts)
print("\nHTTP Status code counts:")
print(status_counts)

# Write results to CSV
write_to_csv(ip_counts, os.path.join(path, "ip_frequency.csv"), "IP", "Frequency")
write_to_csv(status_counts, os.path.join(path, "http_status_frequency.csv"), "HTTP_Status", "Count")

print("\nCSV files created successfully!")
```
