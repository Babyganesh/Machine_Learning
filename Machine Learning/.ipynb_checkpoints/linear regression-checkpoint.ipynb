```python
#split input and output

input_data= car_data .drop(columns=["car_name","selling_price"])
input_data.info()
```

```python
output_data = car_data["selling_price"]
output_data.head()
output_data.info()
```

```python
# dummy data 

input_data= pd.get_dummies(input_data,dtype=int)
print(input_data.head())
```

```python
# for normalise value (standardscal)
#why this using means...data value is not standard 

from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
input_data = scalar.fit_transform(input_data)
print(input_data)
```

```python
from sklearn.model_selection import train_test_split

input_data_train,input_data_test,output_data_train,output_data_test=train_test_split(input_data,output_data,test_size=0.2)
```

```python
    print(input_data_train.shape)
```

```python
    print(input_data_test.shape)
```

```python
# from  sklearn.liner_model import LinerRegression #algrotthim
from sklearn.linear_model import LinearRegression

# to calculate absolute sqr
from sklearn import metrics

object_model = LinearRegression()

object_model.fit(input_data_train, output_data_train)# to train data
```

```python
predicted_selling_price = object_model.predict(input_data_test)
predicted_selling_price
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
# to predict the absolute error and square error
# Remove metrics from both print statements.(need to remove metrics from the print statements!)

print("mean absolute error",mean_absolute_error(predicted_selling_price ,output_data_test))
print("mean squared error",mean_squared_error(predicted_selling_price ,output_data_test))
```
