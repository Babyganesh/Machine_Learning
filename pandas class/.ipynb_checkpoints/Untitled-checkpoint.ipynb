```python
# 3. Add multiple rows
new_rows = pd.DataFrame([
    ["Toyota","White", 60000 , 4 ,"$6,250.00","CNG"],
  ["Nissan","White", 31600 , 4,"$9,700.00","CNG"],
  ["Honda","blue" ,260000,4,"$5,600.00","CNG"]
], columns=new_data.columns)

new_data = pd.concat([new_data, new_rows],ignore_index=True)
# , ignore_index=True

print(new_data)
```

```python
# find duplicate
print("\nDuplicate rows:", new_data.duplicated().sum())
new_data.info()
```

```python
# 3. Remove exact duplicate rows
new_data = new_data.drop_duplicates()
print(new_data)
len(new_data)
```
