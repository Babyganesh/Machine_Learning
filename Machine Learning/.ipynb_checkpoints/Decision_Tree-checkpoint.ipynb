```python
#to divide input data and out put data
input_data = movie_data.drop(columns=['interest'])
input_data
```

```python
output_data = movie_data["interest"]
output_data
```

```python
input_data.describe()
```

```python
output_data.describe()
```

```python
from  sklearn.tree import DecisionTreeClassifier # here decisiontreeclassifier model is class

object_movidemodel = DecisionTreeClassifier()

object_movidemodel .fit(input_data,output_data)
```

```python
movie_data
```

```python
# to predict output (to take object)

object_movidemodel.predict([[34,0],[5,1]])

#just only predict the valuse in givrn or not given data,

    #so need to set accuracy ,
#before split the data like test and train
    
```

```python

from  sklearn.tree import DecisionTreeClassifier

# split the data like test and train
from sklearn.model_selection import train_test_split #here train_test_split is class

#to use import accuracy score (here accuracy_score is function not class)
from sklearn.metrics import accuracy_score

input_data_train,input_data_test,output_data_train,output_data_test = train_test_split (input_data,output_data,test_size=0.2)
object_movidemodel = DecisionTreeClassifier()
object_movidemodel.fit( input_data_train,output_data_train)
```

```python
# to predict efficency ouput

movie_intrestoutput = object_movidemodel.predict(input_data_test)
movie_intrestoutput
```

```python
 # to check accuracy 

accuracy = accuracy_score(output_data_test,movie_intrestoutput) # actual output,predicted output
accuracy
```

```python
movie_intrestoutput = object_movidemodel.predict([[20,1],[13,0]])
movie_intrestoutput
```

**Final concept to remember:**

***Data → Input (X) + Output (Y) → Train Model → Test Model → Predict → Accuracy***
