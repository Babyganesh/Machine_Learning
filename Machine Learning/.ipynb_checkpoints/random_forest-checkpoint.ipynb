```python
output_data = car_data["selling_price"]
output_data.head()
```

```python
print(car_data["fuel_type"].value_counts())
print()
print(car_data["seller_type"].value_counts())
print()
print(car_data["transmission"].value_counts())
print()
```

```python
car_data.replace({"fuel_type":{"Petrol":0,"Diesel":1,"CNG":2}},inplace=True)
```

```python
car_data.replace({"seller_type":{"Dealer":0,"Individual":1}},inplace=True)
```

```python
car_data.replace({"transmission":{"Manual":0,"Automation":1}},inplace=True)
```

```python
car_data.info()
```

```python
#split input and output
input_data= car_data .drop(columns=["car_name","selling_price","transmission"])
input_data.info()
```

```python
output_data = car_data["selling_price"]
output_data.head()
```

```python
# from sklearn.preprocessing import StandardScaler
# scalar = StandardScaler()
# input_data = scalar.fit_transform(input_data)
# print(input_data)
```

```python
from sklearn.model_selection import train_test_split

input_data_train,input_data_test,output_data_train,output_data_test=train_test_split(input_data,output_data,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestRegressor
object_model = RandomForestRegressor()
object_model.fit(input_data_train, output_data_train)# to train data
```

```python
# to find test output for selling
predicted_selling_price = object_model.predict(input_data_test)
predicted_selling_price
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn import metrics
print("mean absolute error",mean_absolute_error(predicted_selling_price ,output_data_test))
```
